```python
import os
import zipfile

project_path = r"C:\Users\saksh\jupyter"
zip_path = os.path.join(project_path, "SentimentScope.zip")

files_to_include = [
    "app.py",
    "requirements.txt",
    "sentiment_model.pkl",
    "tfidf_vectorizer.pkl",
    "templates/index.html",
    "static/style.css",
    "static/script.js"
]

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:
    for file in files_to_include:
        full_path = os.path.join(project_path, file)

        if os.path.exists(full_path):
            zipf.write(full_path, arcname=file)
            print("Added:", file)
        else:
            print("MISSING:", file)

print("\nZIP created at:", zip_path)
```
